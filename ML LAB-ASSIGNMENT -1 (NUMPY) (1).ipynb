```python
#PART 15  (AP24110011428)  # section/comment describing the following code
# Step 4: Bag-of-Words vectors (both using the SAME vocabulary)  # section/comment describing the following code
def bag_of_words_vector(tokens, vocabulary):  # defines a reusable function
    vec = np.array([tokens.count(word) for word in vocabulary])  # creates a NumPy array and stores it
    return vec  # returns the calculated result

vec = bag_of_words_vector(tokens, vocabulary)  # assigns the calculated value to `vec`
vec2 = bag_of_words_vector(tokens2, vocabulary)  # assigns the calculated value to `vec2`

print("Vector A:", vec)  # displays the result
print("Vector B:", vec2)  # displays the result
```

```python
#PART 15  (AP24110011428)  # section/comment describing the following code
# Step 5: Cosine Similarity  # section/comment describing the following code
import numpy as np  # imports the required Python module
def cosine_similarity(vec1, vec2):  # defines a reusable function
    dot_product = np.dot(vec1, vec2)  # calculates the dot product
    norm1 = np.linalg.norm(vec1)  # calculates the vector norm
    norm2 = np.linalg.norm(vec2)  # calculates the vector norm
    return dot_product / (norm1 * norm2)  # returns the calculated result

similarity = cosine_similarity(vec, vec2)  # assigns the calculated value to `similarity`
print("Cosine similarity:")  # displays the result
print(similarity)  # displays the result
print("So approximately:")  # displays the result
print(f"Cosine similarity: {similarity:.4f}")  # displays the result
```

Cosine similarity compares the direction of two vectors rather than simply comparing their sizes.
```python
#NLP Mini Project: Text Similarity Checker  # section/comment describing the following code
import numpy as np  # imports the required Python module

def bag_of_words_vector(tokens, vocabulary):  # defines a reusable function
    vector = np.array([  # creates a NumPy array and stores it
        np.sum(np.array(tokens) == word)  # creates a NumPy array and stores it
        for word in vocabulary  # iterates through the selected items
    ])  # executes the specified operation
    return vector  # returns the calculated result


# Three sentences  # section/comment describing the following code
sentence1 = "I enjoy learning Python"  # assigns the calculated value to `sentence1`
sentence2 = "I like learning Python"  # assigns the calculated value to `sentence2`
sentence3 = "The football match was exciting"  # assigns the calculated value to `sentence3`

# Tokenize  # section/comment describing the following code
tokens1 = sentence1.lower().split()  # converts the text to lowercase
tokens2 = sentence2.lower().split()  # converts the text to lowercase
tokens3 = sentence3.lower().split()  # converts the text to lowercase

# Shared vocabulary  # section/comment describing the following code
vocabulary = np.unique(  # finds the unique values
    np.array(tokens1 + tokens2 + tokens3)  # executes the specified operation
)  # executes the specified operation

# Bag-of-words vectors  # section/comment describing the following code
vec1 = bag_of_words_vector(tokens1, vocabulary)  # assigns the calculated value to `vec1`
vec2 = bag_of_words_vector(tokens2, vocabulary)  # assigns the calculated value to `vec2`
vec3 = bag_of_words_vector(tokens3, vocabulary)  # assigns the calculated value to `vec3`


def cosine_similarity(a, b):  # defines a reusable function
    return np.dot(a, b) / (  # returns the calculated result
        np.linalg.norm(a) * np.linalg.norm(b)  # calculates the vector norm
    )  # executes the specified operation


# Similarities  # section/comment describing the following code
sim12 = cosine_similarity(vec1, vec2)  # assigns the calculated value to `sim12`
sim13 = cosine_similarity(vec1, vec3)  # assigns the calculated value to `sim13`
sim23 = cosine_similarity(vec2, vec3)  # assigns the calculated value to `sim23`

print("Similarity 1-2:", sim12)  # displays the result
print("Similarity 1-3:", sim13)  # displays the result
print("Similarity 2-3:", sim23)  # displays the result

similarities = {  # assigns the calculated value to `similarities`
    "Sentence 1 and Sentence 2": sim12,  # executes the specified operation
    "Sentence 1 and Sentence 3": sim13,  # executes the specified operation
    "Sentence 2 and Sentence 3": sim23  # executes the specified operation
}  # executes the specified operation

most_similar = max(similarities, key=similarities.get)  # assigns the calculated value to `most_similar`

print("Most similar pair:", most_similar)  # displays the result
```
